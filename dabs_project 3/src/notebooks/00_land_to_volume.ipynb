```python
dbutils.widgets.text("environment_name", "")
dbutils.widgets.text("volume_path", "/Volumes/delta_lake_catalog/default/deltalake_volume")
dbutils.widgets.text("source_table", "samples.tpch.orders")
```

```python
env_name = dbutils.widgets.get("environment_name")
volume_path = dbutils.widgets.get("volume_path")
source_table = dbutils.widgets.get("source_table")

raw_landing_path = f"{volume_path}/raw/orders"

print(f"environment name is {env_name}")
print(f"reading from Databricks sample table: {source_table}")
print(f"landing raw csv files to: {raw_landing_path}")
```

```python
# This step simulates a "raw file drop" - pulling from a Databricks-provided
# sample dataset and landing it as CSV in a Volume, exactly like a source system
# would drop files for ingestion in a real pipeline.

source_df = spark.table(source_table)

(
    source_df
        .write
        .mode("overwrite")
        .option("header", "true")
        .csv(raw_landing_path)
)

print(f"Landed {source_df.count()} rows to {raw_landing_path}")
```
